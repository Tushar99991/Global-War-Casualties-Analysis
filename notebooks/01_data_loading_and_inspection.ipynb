```python
import pandas as pd
import numpy as np
from pathlib import Path
import os

from pathlib import Path

# Go one level up from notebooks/ to the real project root
PROJECT_ROOT = Path.cwd().parent
RAW_DATA = PROJECT_ROOT / "data" / "raw"
PROCESSED_DATA = PROJECT_ROOT / "data" / "processed"

print("Project root :", PROJECT_ROOT)
print("Raw data path:", RAW_DATA)
print("Exists?      :", RAW_DATA.exists())

# --------------------------------------------------
# 2. List all files in raw folder
# --------------------------------------------------
print("=" * 60)
print("FILES IN data/raw/")
print("=" * 60)

if RAW_DATA.exists():
    files = list(RAW_DATA.glob("*"))
    if files:
        for f in sorted(files):
            print(f"  - {f.name}")
    else:
        print("  (folder is empty)")
else:
    print("  ERROR: data/raw folder not found!")

print()

# --------------------------------------------------
# 3. Load datasets
# --------------------------------------------------
print("=" * 60)
print("LOADING DATASETS")
print("=" * 60)

datasets = {}

# Helper function
def load_csv(filename, name):
    path = RAW_DATA / filename
    try:
        df = pd.read_csv(path, low_memory=False)
        print(f"✓ {filename:45} → {df.shape[0]:>7,} rows × {df.shape[1]} columns")
        datasets[name] = df
        return df
    except Exception as e:
        print(f"✗ {filename:45} → {e}")
        datasets[name] = None
        return None

# Load each file
owid          = load_csv("owid_war_deaths_raw.csv", "OWID")
ucdp          = load_csv("ucdp_ukraine_russia_raw.csv", "UCDP")
gaza_daily    = load_csv("techforpalestine_gaza_daily_raw.csv", "Gaza Daily")
westbank      = load_csv("techforpalestine_westbank_daily_raw.csv", "West Bank Daily")
killed_gaza   = load_csv("killed_in_gaza.csv", "Killed in Gaza")

print()

# --------------------------------------------------
# 4. Detailed inspection of each loaded dataset
# --------------------------------------------------
for name, df in datasets.items():
    if df is not None:
        print("=" * 60)
        print(f"{name}")
        print("=" * 60)
        print(f"Shape        : {df.shape[0]:,} rows × {df.shape[1]} columns")
        print(f"Columns      : {list(df.columns)}")
        print("\nFirst 5 rows:")
        print(df.head())
        print("\nData types & missing values:")
        info = pd.DataFrame({
            "Column": df.columns,
            "Dtype": df.dtypes.values,
            "Non-Null": df.count().values,
            "Nulls": df.isnull().sum().values,
            "Null %": (df.isnull().sum() / len(df) * 100).round(2).values
        })
        print(info.to_string(index=False))
        print("\n")
```

```python
print("\n" + "="*60)
print("EXACT COLUMN NAMES")
print("="*60)

for name, df in datasets.items():
    if df is not None:
        print(f"\n{name}:")
        print(list(df.columns))
```
